```python
df.drop(
    columns=[
        'EmployeeCount',
        'EmployeeNumber',
        'Over18',
        'StandardHours'
    ],
    inplace=True
)
```

```python
df.shape
```

```python
attrition_rate = (
    df['Attrition']
    .value_counts(normalize=True)['Yes']
    * 100)
print(f"Attrition Rate: {attrition_rate:.2f}%")
```

### The organization has an attrition rate of X%, meaning X out of every 100 employees leave the company.

```python
sns.countplot(x='Attrition', data=df)
plt.title('Employee Attrition Distribution')
plt.show()
```

```python
dept_attrition = pd.crosstab(df['Department'],df['Attrition'])
dept_attrition
```

```python
dept_attrition = pd.crosstab(df['Department'],df['Attrition']).plot(kind='bar', figsize=(8,5))
plt.xticks(rotation=0)
plt.title('Department-wise Attrition')
plt.show()
```

```python
attrition_rate = (df['Attrition'].value_counts(normalize=True)['Yes']* 100)
print(attrition_rate)
```

```python
overtime_attrition = pd.crosstab(df['OverTime'],df['Attrition'],normalize='index') * 100
overtime_attrition
```

```python
sns.countplot(x='OverTime',hue='Attrition',data=df)
plt.title('Overtime vs Attrition')
plt.show()
```

### Employees working overtime show significantly higher attrition rates, suggesting workload may be a key factor driving employee turnover.

```python
df.groupby('Attrition')['MonthlyIncome'].mean()
```

```python
sns.boxplot(x='Attrition',y='MonthlyIncome',data=df)
plt.title('Salary Distribution by Attrition')
plt.show()
```

### Employees who left the organization generally earned lower salaries than employees who stayed.

```python
pd.crosstab(df['JobSatisfaction'],df['Attrition'],normalize='index') * 100
```

```python
sns.countplot(x='JobSatisfaction',hue='Attrition',data=df)
plt.title('Job Satisfaction vs Attrition')
plt.show()
```

### Employees with lower job satisfaction scores exhibit a higher attrition rate compared to employees with higher satisfaction scores.

```python
pd.crosstab(df['WorkLifeBalance'],df['Attrition'],normalize='index') * 100
```

```python
sns.countplot(x='WorkLifeBalance',hue='Attrition',data=df)
plt.title('Work Life Balance vs Attrition')
plt.show()
```

### Poor work-life balance appears to be associated with increased employee turnover.

```python
df.groupby('Attrition')['YearsAtCompany'].mean()
```

```python
sns.boxplot(x='Attrition',y='YearsAtCompany',data=df)
plt.title('Years at Company vs Attrition')
plt.show()
```

### Employees who leave generally have fewer years of service than those who stay, indicating attrition is more common among newer employees.

```python
job_attrition = pd.crosstab(df['JobRole'],df['Attrition'])
job_attrition
```

```python
job_attrition.plot(kind='bar',figsize=(12,6))
plt.title('Job Role vs Attrition')
plt.xticks(rotation=30)
plt.show()
```

### Certain job roles experience significantly higher turnover and may require targeted retention strategies.

```python
numeric_df = df.select_dtypes(include=['int64','float64'])
plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(),cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```
